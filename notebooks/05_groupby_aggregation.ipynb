# Day 5 — Grouping and Aggregation

This notebook explores condition-level patterns in a synthetic XR moral decision-making dataset using Pandas grouping and aggregation tools.

Topics covered:
- Grouping data with `groupby()`
- Calculating group-level summary statistics
- Applying multiple aggregation functions
- Counting categorical outcomes within groups
- Comparing behavioral measures across experimental conditions

The goal is to move from participant-level observations to condition-level summaries that can later support statistical analysis.

```python
import pandas as pd
```

```python
df = pd.read_csv("/content/vr_moral_dilemma_clean.csv")
```

```python
cond_grp = df.groupby("condition")
```

```python
cond_grp["decision_time"].mean()
```

```python
cond_grp["embodiment_score"].mean()
```

```python
cond_grp["conflict_score"].mean()
```

```python
df["condition"].value_counts()
```

```python
cond_grp["decision_time"].agg(["mean","median", "std"] )
```

> The sheep and farmer conditions showed relatively similar mean decision times, although the farmer condition exhibited greater variability, as indicated by a larger standard deviation. In contrast, the sheep condition showed a higher mean conflict score than the farmer condition.

```python
cond_grp[["decision_time", "conflict_score"]].agg(["mean", "std"])
```

```python
cond_grp["choice"].value_counts()
```

```python
summary_table = cond_grp.agg(
    decision_time_mean=("decision_time", "mean"),
    decision_time_std=("decision_time", "std"),
    conflict_mean=("conflict_score", "mean")
)

summary_table
```

````python
clean_df = df.copy()
````

````python
clean_df = clean_df.drop_duplicates()
````

````python
clean_df = clean_df.dropna()
````

````python
clean_df.isna().sum()
````

````python
clean_df = clean_df[
    clean_df["embodiment_score"].between(1, 7)
    & clean_df["presence_score"].between(1, 7)
    & clean_df["agency_score"].between(1, 7)
    & clean_df["conflict_score"].between(1, 7)
]
````

````python
condition_grp = df.groupby(["condition"])
````

````python
condition_grp.get_group("sheep")
````

````python
condition_grp[["decision_time","presence_score","conflict_score","embodiment_score","agency_score"]].agg(["mean","median","std"])
````

````python
summary_table = condition_grp.agg(
    n=('participant_id', 'count'),
    decision_time_mean=('decision_time', 'mean'),
    decision_time_std=('decision_time', 'std'),
    conflict_mean=('conflict_score', 'mean'),
    embodiment_mean=('embodiment_score', 'mean'),
    gaze_switches_mean=('gaze_switches', 'mean')
)
display(summary_table)
````

````python
choice_behavior = condition_grp["choice"].value_counts(normalize=True)
print(choice_behavior)
````

the first fixation behavior looks descrpitively relavent between the condition and the first fixation in terms of embodiment manipulation

````python
first_fixation_beh = condition_grp["first_fixation"].value_counts(normalize = True)
print(first_fixation_beh)
````

````python
filtered_df = clean_df[clean_df['choice'] == 'save_sheep']

# Group by 'condition' and calculate the mean of the specified columns
comparison_results = filtered_df.groupby('condition')[['decision_time', 'conflict_score', 'gaze_time_sheep']].mean()

display(comparison_results)
````
